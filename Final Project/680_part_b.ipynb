```python
from PIL import Image, ImageEnhance
import numpy as np
import os
import pandas as pd

labels_df = pd.read_csv(prepared_labels_path)
IMAGE_COL = "image_path"

class_counts = labels_df['label'].value_counts()
max_count = class_counts.max()

print("Class distribution before augmentation:")
print(class_counts)

def augment_image(img):
    augmented_versions = []

    # rotate
    for angle in [90, 180, 270]:
        augmented_versions.append(img.rotate(angle))

    # flips
    augmented_versions.append(img.transpose(Image.FLIP_LEFT_RIGHT))
    augmented_versions.append(img.transpose(Image.FLIP_TOP_BOTTOM))

    # brightness adjustments
    for factor in [0.7, 1.3]:
        augmented_versions.append(ImageEnhance.Brightness(img).enhance(factor))

    # contrast adjustments
    for factor in [0.7, 1.3]:
        augmented_versions.append(ImageEnhance.Contrast(img).enhance(factor))

    return augmented_versions

augmented_metadata = []

for label_class, count in class_counts.items():
    needed = max_count - count
    subset = labels_df[labels_df['label'] == label_class]

    if needed > 0:
        print(f"Oversampling {label_class}: need {needed} more samples")

        for i in range(needed):
            row = subset.sample(1).iloc[0]

            img_filename = row[IMAGE_COL]  # <-- now uses 'image_path'
            img_path = os.path.join(prepared_dir, img_filename)

            img = Image.open(img_path)

            aug_images = augment_image(img)
            aug_img = aug_images[np.random.randint(len(aug_images))]

            new_filename = f"{os.path.splitext(img_filename)[0]}_aug_{i}.jpg"
            new_path = os.path.join(augmented_dir, new_filename)

            aug_img.save(new_path)

            augmented_metadata.append([new_filename, label_class])

augmented_df = pd.DataFrame(augmented_metadata, columns=[IMAGE_COL, 'label'])
augmented_df.to_csv(os.path.join(project_path, "augmented_labels.csv"), index=False)

print("Augmentation complete — balanced dataset generated!")
```

```python
import os
import pandas as pd

aug_dir = os.path.join(project_path, "augmented_images")
orig_labels_path = prepared_labels_path

orig_labels = pd.read_csv(orig_labels_path)   # has image_path + label

aug_files = sorted([
    f for f in os.listdir(aug_dir)
    if f.lower().endswith(".jpg") and "_aug_" in f
])

aug_df = pd.DataFrame({"image_path_aug": aug_files})

def to_base_name(fn: str) -> str:
    base = fn.split("_aug_")[0]
    return base + ".jpg"

aug_df["base_image_path"] = aug_df["image_path_aug"].apply(to_base_name)

merged = pd.merge(
    aug_df,
    orig_labels[["image_path", "label"]],
    left_on="base_image_path",
    right_on="image_path",
    how="left"
)

final_aug_labels = merged[["image_path_aug", "label"]].rename(
    columns={"image_path_aug": "image_path"}
)

# new name
out_path = os.path.join(project_path, "AUG_ONLY_WITH_LABELS_DO_NOT_EDIT.csv")
final_aug_labels.to_csv(out_path, index=False)

print("Wrote file to:", out_path)
print(final_aug_labels.head())




```

Lesion Label Meanings (Legend)

1. akiec
Full term: Actinic Keratoses and Intraepithelial Carcinoma (Bowen’s Disease)
Type: Precancerous / early-stage skin cancer
Notes: Can progress to squamous cell carcinoma
2. bcc
Full term: Basal Cell Carcinoma
Type: Skin cancer
Notes: Most common skin cancer, locally invasive but rarely spreads
3. bkl
Full term: Benign Keratosis-like Lesions
Includes:
Seborrheic keratosis
Solar lentigo
Lichen-planus–like keratosis
Type: Benign
Notes: Often mistaken for melanoma visually
4. df
Full term: Dermatofibroma
Type: Benign fibrous nodule
Notes: Usually small, firm nodules on extremities
5. mel
Full term: Melanoma
Type: Malignant (dangerous skin cancer)
Notes: Can metastasize, highest clinical importance
6. nv
Full term: Melanocytic Nevus
Type: Mole / benign melanocytic proliferation
Notes: Most common lesion type in dataset
7. vasc
Full term: Vascular Lesions
Includes:
Hemangiomas
Angiokeratoma
Pyogenic granuloma
Hematoma
Type: Usually benign
Notes: Red or purple appearance
